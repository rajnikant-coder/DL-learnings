```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score , classification_report,confusion_matrix
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.layers import Flatten
from tensorflow.keras.layers import MaxPooling2D
from tensorflow.keras.layers import Dropout
from tensorflow.keras.utils import to_categorical
```

```python
df = pd.read_csv("mnist_train.csv")
df_test = pd.read_csv ("mnist_test.csv")
```

```python
#preprocess 


x_train = df.drop("label" , axis = 1).values
y_train = df["label"].values
x_test = df_test.drop("label" , axis = 1).values
y_test = df_test["label"].values
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
x_train_img = x_train.reshape(-1,28,28)
x_test_img = x_test.reshape(-1,28,28)
```

```python
x_train_img
```

```python
y_train_cat = to_categorical(y_train ,10)
y_test_cat = to_categorical(y_test ,10)
```

# perceptron

```python
model_perceptron = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(10, activation = "softmax")
])
```

```python
model_perceptron.compile(optimizer = "sgd" , loss = "categorical_crossentropy" , metrics = ["accuracy"])
```

```python
history_percp = model_perceptron.fit(
    x_train_img , y_train_cat ,
    epochs = 5 ,
    batch_size = 32 , 
    validation_data =(x_test_img , y_test_cat) , 
    verbose = 1
    )
```

```python
acc_percp = model_perceptron.evaluate(x_test_img , y_test_cat , verbose = 1)
acc_percp
```

# ANN

```python
model_ann= Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128, activation = "relu"),
    Dense(64, activation = "relu"),
    Dense(10, activation = "softmax"),
])
```

```python
model_ann.compile(optimizer = "adam" , loss = "categorical_crossentropy" , metrics = ["accuracy"])
```

```python
history_ann = model_ann.fit(
    x_train_img , y_train_cat , 
    epochs = 5 ,
    batch_size = 32 , 
    validation_data =(x_test_img , y_test_cat) , 
    verbose = 1
    )
```

```python
acc_ann = model_ann.evaluate(x_test_img , y_test_cat , verbose = 1)[1]
acc_ann
```

# CNN

```python

x_train_cnn = x_train.reshape(-1,28,28,1)
x_test_cnn = x_test.reshape(-1,28,28,1)
```

```python
model_cnn = Sequential([
    Conv2D(32, kernel_size = (3,3) ,activation = "relu",input_shape = (28,28 ,1)),
    MaxPooling2D(pool_size = (2,2)),
    Conv2D(64, kernel_size = (3,3) ,activation = "relu"),
    MaxPooling2D(pool_size = (2,2)),
    Flatten(),
    Dense(128, activation = "relu"),
    Dropout(0.5),
    Dense(10, activation = "softmax"),
])
```

```python
model_cnn.compile(optimizer="adam" , loss = "categorical_crossentropy" , metrics=["accuracy"])
```

```python
history_cnn = model_cnn.fit(
    x_train_cnn , y_train_cat , 
    epochs = 5 ,
    batch_size = 32 , 
    validation_data =(x_test_cnn , y_test_cat) , 
    verbose = 1
    )
```

```python
acc_cnn = model_cnn.evaluate(x_test_cnn, y_test_cat , verbose = 1)[1]
acc_cnn
```
